```python
##### import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
from scipy.optimize import curve_fit

def analyze_nmr_final(file_path, label, viscosity_cP, G=0.1, temp_K=295):
    # 1. Physics Constants
    k_B, r = 1.3806e-23, 2.0e-10
    eta = viscosity_cP * 1e-3
    D = (k_B * temp_K) / (6 * np.pi * eta * r)
    gamma = 2 * np.pi * 15.53e6  # rad s^-1 T^-1 (Hardware Specific)
    
    # 2. Load & Clean Data (Handling MS and Multi-Column structure)
    try:
        # Skip 2 lines to get past "Time, Channel A" and "(ms), (V)"
        df = pd.read_csv(file_path, encoding='windows-1252', skiprows=2, usecols=[0, 1], names=['t_ms', 'v'])
        # Force conversion to numeric to fix the "str / float" error
        df['t_ms'] = pd.to_numeric(df['t_ms'], errors='coerce')
        df['v'] = pd.to_numeric(df['v'], errors='coerce')
        df = df.dropna().reset_index(drop=True)
        df['t_s'] = df['t_ms'] / 1000.0
    except Exception as e:
        print(f"Skipping {label}: {e}")
        return None

    # 3. Pulse-to-Echo Detection
    # Detect pulses (the big spikes ~1V)
    # Distance=40 (~4ms) ensures we don't double-count a single pulse
    pulses, _ = find_peaks(df['v'], height=0.5, distance=40)
    t_pulses = df.loc[pulses, 't_s'].values
    
    if len(t_pulses) < 5:
        print(f"Skipping {label}: Pulse sequence not recognized.")
        return None
    
    # Find Echoes: Look for the maximum voltage between each pulse
    echo_t, echo_v = [], []
    for i in range(len(t_pulses) - 1):
        t_start, t_end = t_pulses[i], t_pulses[i+1]
        # Look in the middle 60% of the window to avoid the pulse ringing
        mask = (df['t_s'] > t_start + 0.001) & (df['t_s'] < t_end - 0.001)
        window = df[mask]
        if not window.empty:
            idx = window['v'].idxmax()
            echo_t.append(window.loc[idx, 't_s'])
            echo_v.append(window.loc[idx, 'v'])

    # Filter out bad echoes (must be above noise floor and follow decay style)
    echo_t, echo_v = np.array(echo_t), np.array(echo_v)
    valid_mask = (echo_v > 0.015) & (echo_v < 0.6) # Ignores noise and pulses
    t_fit_raw = echo_t[valid_mask]
    v_fit_raw = echo_v[valid_mask]
    
    if len(t_fit_raw) < 5: return None

    # 4. Diffusion-Corrected Fitting
    t_zeroed = t_fit_raw - t_fit_raw[0]
    te = np.median(np.diff(t_pulses))
    tau = te / 2
    K = (gamma**2 * G**2 * D * tau**2) / 3

    def decay_model(t, A, T2_act, C):
        return A * np.exp(-t * (1.0/T2_act + K)) + C

    try:
        popt, _ = curve_fit(decay_model, t_zeroed, v_fit_raw, p0=[v_fit_raw[0], 0.2, 0.01], bounds=(0, [2, 10, 0.5]))
        T2_act = popt[1]
    except: return None

    # 5. Plotting Individual Result
    plt.figure(figsize=(6, 4))
    plt.semilogy(t_zeroed, v_fit_raw, 'bo', alpha=0.5, label='Measured Echoes')
    plt.semilogy(t_zeroed, decay_model(t_zeroed, *popt), 'r-', label=f'Fit T2={T2_act:.4f}s')
    plt.title(f"{label} Glycerol Decay")
    plt.legend()
    plt.grid(True, which='both', alpha=0.2)
    plt.show()

    return {"Concentration": label, "Viscosity (cP)": viscosity_cP, "T2 (s)": round(T2_act, 5)}

# --- Full File Path Configuration ---
configs = [
    {"path": r"C:\Users\Student\OneDrive - University of Cambridge\Desktop\E2\Glycerol\T2\Gly 1.5.csv", "visc": 1.05, "name": "1.5%"},
    {"path": r"C:\Users\Student\OneDrive - University of Cambridge\Desktop\E2\Glycerol\T2\Gly 3.05.csv", "visc": 1.09, "name": "3.05%"},
    {"path": r"C:\Users\Student\OneDrive - University of Cambridge\Desktop\E2\Glycerol\T2\Gly 6.1.csv", "visc": 1.18, "name": "6.1%"},
    {"path": r"C:\Users\Student\OneDrive - University of Cambridge\Desktop\E2\Glycerol\T2\Gly 12.5.csv", "visc": 1.41, "name": "12.5%"},
    {"path": r"C:\Users\Student\OneDrive - University of Cambridge\Desktop\E2\Glycerol\T2\Gly 50.csv", "visc": 6.00, "name": "50%"},
    {"path": r"C:\Users\Student\OneDrive - University of Cambridge\Desktop\E2\Glycerol\T2\Gly 75.csv", "visc": 35.5, "name": "75%"},
    {"path": r"C:\Users\Student\OneDrive - University of Cambridge\Desktop\E2\Glycerol\T2\Gly 90.csv", "visc": 219.0, "name": "90%"},
    {"path": r"C:\Users\Student\OneDrive - University of Cambridge\Desktop\E2\Glycerol\T2\Gly 100.csv", "visc": 1410.0, "name": "100%"}
]

# Run analysis
final_data = []
for c in configs:
    res = analyze_nmr_final(c['path'], c['name'], c['visc'])
    if res: final_data.append(res)

# Print Summary Table
print("\n--- FINAL T2 ANALYSIS SUMMARY ---")
print(pd.DataFrame(final_data).to_string(index=False))
```
